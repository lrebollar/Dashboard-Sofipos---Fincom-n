```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np
import pickle
import os
```

```python
with open('dict_sofipos_sector.pkl', 'rb') as archivo:
    dict_sofipos = pickle.load(archivo)
```

```python
dict_sofipos['fintech'].keys()
```

```python
df_indf_fintech = dict_sofipos['fintech']['Inf_cartera']
df_pasivos_fintech = dict_sofipos['fintech']['Capt_trad']
```

```python
df_pasivos_fintech
```

```python
df = pd.merge(left=df_indf_fintech[['periodo', 'Activo', 'sofipo']], right=df_pasivos_fintech[['periodo', 'Pasivo', 'sofipo']], on=['sofipo', 'periodo'], how='left')
```

```python
df
```

```python
df['Activo'] = df['Activo']/1000000
df['Pasivo'] = df['Pasivo']/1000000
df['periodo'] = pd.to_datetime(df['periodo'], format='%Y-%m')
```

```python
v_periodo = df[df['sofipo'] == 'Fincomún']['periodo']
v_act_finco = df[df['sofipo'] == 'Fincomún']['Activo']
v_pasv_finco = df[df['sofipo'] == 'Fincomún']['Pasivo']

fig = go.Figure()

fig.add_trace(go.Bar(
    x=v_periodo,
    y=v_act_finco,
    name="Activo"
))

fig.add_trace(go.Bar(
    x=v_periodo,
    y=v_pasv_finco,
    name="Pasivo"
))

fig.update_layout(
    barmode="group",  # "group" para barras lado a lado, "stack" para apiladas
    title="Comparación de dos series de tiempo",
    xaxis_title="Fecha",
    yaxis_title="Valor",
    xaxis=dict(tickformat="%b %Y"),
    legend_title="Series",
    template="plotly_white"
)

fig.show()
```

```python
df_pasivo = df_pasivos_fintech[['periodo', 'Pasivo', 'Captación tradicional', 'sofipo']]
df_pasivo['periodo'] = pd.to_datetime(df_pasivo['periodo'], format='%Y-%m')
df_pasivo = df_pasivo[df_pasivo['periodo'] >= '2024-01-01']

df_pasivo['Pasivo'] = df_pasivo['Pasivo']/1000000
df_pasivo['Captación tradicional'] = df_pasivo['Captación tradicional']/1000000
```

```python
v_periodo = df_pasivo[df_pasivo['sofipo'] == 'Fincomún']['periodo']
v_captacion_finco = df_pasivo[df_pasivo['sofipo'] == 'Fincomún']['Captación tradicional']
v_pasv_finco = df_pasivo[df_pasivo['sofipo'] == 'Fincomún']['Pasivo']
v_ind = (v_pasv_finco/v_captacion_finco)*100

fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
    go.Bar(
        x=v_periodo,
        y=v_captacion_finco,
        name=f"Captación Tradicional ({v_captacion_finco.iloc[-1]:,.1f})"
    ),
    secondary_y=False
)

fig.add_trace(
    go.Bar(
        x=v_periodo,
        y=v_pasv_finco,
        name=f"Pasivo ({v_pasv_finco.iloc[-1]:,.1f})"
    ),
    secondary_y=False
)

fig.add_trace(
    go.Scatter(
        x=v_periodo,
        y=v_ind,
        name=f"Pasivo / Captación ({v_ind.iloc[-1]:.1f}%)",
        mode="lines+markers",
        line=dict(color="black", width=2)
    ),
    secondary_y=True
)

fig.update_layout(
    barmode="group",
    title="Captación/Pasivo - Fincomún",
    xaxis_title="Fecha",
    legend=dict(
        x=0.01,
        y=0.99,
        xanchor="left",
        yanchor="top",
        bgcolor="rgba(255,255,255,0.6)",
        bordercolor="rgba(0,0,0,0.2)",
        borderwidth=1
    ),
    template="plotly_white"
)

fig.update_xaxes(tickformat="%b %Y")
fig.update_yaxes(title_text="Monto en MDP ($)", secondary_y=False)
fig.update_yaxes(title_text="Pasivo / Captación (%)", secondary_y=True)

fig.show()
```
